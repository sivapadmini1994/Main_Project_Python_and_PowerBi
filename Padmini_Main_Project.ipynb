## Stage 2 – Data Cleaning and Pre-processing

- Handle missing values (impute or drop)

- Handle duplicates
- column name rename or dropping unwanted columns if required

- Convert data types if needed

```python
# your code here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
# Handling missing values and Handling Duplicates:
print("\nMising values:", df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())

df.drop_duplicates(inplace=True)

num_cols = df.select_dtypes(include=['number']).columns
cat_cols = df.select_dtypes(include=['object']).columns
for col in num_cols:
  df[col] = df[col].fillna(df[col].mean())
for col in cat_cols:
  df[col] = df[col].fillna(df[col].mode()[0])
print("\nAfter cleaning -> shape:", df.shape)
```

```python
# your code here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
# Column rename and drop unwanted column:
print(df.columns.tolist())
# Clean the column names - Lowercase, Underscore
df.columns = df.columns.str.strip().str.lower().str.replace(' ', '_')
print(df.columns.tolist())

# Drop Booking_ID -> unique identifier, not useful for analysis
df = df.drop(columns=['booking_id'])

df.head()
```

```python
# your code here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
# Treat outliers if required:
import numpy as np
outlires_cols = ["lead_time", "average_price"]

for col in outlires_cols:
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR
  before = df.shape[0]
  df = df[(df[col] >= lower) & (df[col] <= upper)]
  print(f"{col}: removed {before - df.shape[0]} outlier rows")

print("\nShape after outlier removal:", df.shape)
```

```python
# your code here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
# Check skewness and apply transformations:
num_cols = df.select_dtypes(include=['number']).columns
skew_vals = df[num_cols].skew().sort_values(ascending=False)
print("Skewness of numeric columns:\n", skew_vals)

import numpy as np
for col in num_cols:
    if abs(df[col].skew()) > 1 and (df[col] >= 0).all():
        df[col + "_log"] = np.log1p(df[col])
print("\nFinal shape after transformation:", df.shape)

```

```python
# your code here
# If you want more cells, you can add or less cells you can remove, but dont change the structure.
# Convert datatypes and Feature transformations-date parts:
# Convert date of reservation to datetime
df["date_of_reservation"] = pd.to_datetime(df["date_of_reservation"], errors="coerce")
# date parts:
df["reservation_year"] = df["date_of_reservation"].dt.year
df["reservation_month"] = df["date_of_reservation"].dt.month
df["reservation_day"] = df["date_of_reservation"].dt.day
df["reservation_weekday"] = df["date_of_reservation"].dt.day_name()

# Convert booking_status to numeric flag(0 - Not_Canceled, 1 = Canceled)
df["booking_status_flag"] = df["booking_status"].apply(lambda x: 1 if x == "Canceled" else 0)
df[["date_of_reservation", "reservation_year", "reservation_month", "reservation_weekday", "booking_status_flag"]].head()


```

## Key Insights (Trends, Anomalies, Correlations)


1.   ** Lead time is the strongest predictor of cancellation** (correlation 0.41) - cancellations rise sharply once lead time crosses ~150 days, while last-minute bookings are almost always honoured.

2.   ** Online bookings cancel the most** (35.9%), while corporate bookings cancel the least (10.9%).

3.   ** Special requests reduce cancellations** (correlation ~0.25%) - engaged guests cancel less often.

4.   ** Cancelled bookings are priced slightly higher** on average (~$107 median) than honoured bookings (~$95 median).
5.   ** Room_Type 1 drives most cancellations**, peaking seasonally around months 8-10.



## Recommendations for Business / Decision-Making


1.   Introduce a deposit or non-refundable policy for online bookings made more than 150 days in advance.

2.   Encourage special requests or paid add-ons at booking time - engaged guests cancel far less.

3.  Send targeted confirmation/reminder messages to long-lead-time bookings as arrival approaches.

4.  Review pricing/availability for Room_Type 1; Consider dynamic pricing or mild overbooking.

5.  Plan staffing/inventory around the August-October cancellation peak.
